```python
# dependecies
import pandas as pd
import numpy as np
from ctgan import CTGAN
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv('personalFinanceDataset.csv')
```

```python
data = data.drop(['Unnamed: 0'], axis=1)
data.head()
```

```python
catCols = data.select_dtypes(include=['object', 'boolean']).columns.tolist()
```

```python
ctgan = CTGAN(epochs=50)
ctgan.fit(data, catCols)
```

```python
# create synthetic data
synData = ctgan.sample(3500)
```

```python
synData.head()
```

```python
synData.to_csv('synData.csv', index=False)
```

```python
# biggest indication that datasets were different was mean differences; checking that now
numericCols = data.select_dtypes(['number']).columns.tolist()
meanDiffs = []
for col in numericCols:
    ogMean = data[col].mean()
    synMean = synData[col].mean()
    absDiff = abs(ogMean - synMean)
    meanDiffs.append({'Feature': col, 'Mean Difference': absDiff})

print(meanDiffs)
```
